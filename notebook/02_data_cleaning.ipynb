## Module 2: Data Cleaning

### Objective
Clean the raw retail dataset by handling missing values, duplicates, incorrect data types, and invalid records before analysis.

```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt
import seaborn as sns 

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 20)
```

```python
df = pd.read_csv("../data/raw/online_retail_data.csv")
```

```python
df.head()
```

```python
df_clean = df.copy()
```

```python
df_clean.isnull().sum()
```

```python
missing = pd.DataFrame({"missing values": df_clean.isnull().sum(),"percentage": round((df_clean.isnull().sum()/len(df_clean))*100, 2)})

missing
```

```python
df_clean.duplicated().sum()
```

```python
df_clean = df_clean.drop_duplicates()
```

```python
df_clean.duplicated().sum()
```

```python
df_clean.shape
```

```python
df_clean.isnull().sum
```

```python
df_clean["InvoiceDate"] = pd.to_datetime(df_clean["InvoiceDate"])
df_clean["InvoiceDate"].dtype
```

```python
(df_clean["Quantity"] < 0).sum()
(df_clean["Price"] <= 0).sum()
```

```python
cancelled_orders = df_clean[
    df_clean["Invoice"].astype(str).str.startswith("C")
]

cancelled_orders.head()
```

```python
df_clean.to_csv(
    "../data/cleaned/cleaned_online_retail.csv",
    index=False
)
```

```python
df_clean[df_clean["Customer ID"].isnull()]
```

```python
df_clean["Customer ID"].isnull().sum()
```

```python
df_clean = df_clean.dropna(subset=["Customer ID"])
```

```python
df_clean["Customer ID"].isnull().sum()
```

```python
df_clean = df_clean.dropna(subset=["Description"])
```

```python
cancelled_orders = df_clean[
    df_clean["Invoice"].astype(str).str.startswith("C")]

cancelled_orders.shape
```

```python
df_clean = df_clean[
    ~df_clean["Invoice"].astype(str).str.startswith("C")]
```

```python
df_clean[
    df_clean["Invoice"].astype(str).str.startswith("C")].shape
```

```python
df_clean.isnull().sum()
```
